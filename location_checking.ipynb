```python
addy = pd.read_csv("boston_50_addresses.csv")
addy
```

```python
## Run tests of all the addresses and get all the responses stored in a csv file.
data = {"Address": [], "Result": []}
for address in addy["Address"]:
    """
    Create chatbot instance using chat.py
    """
    chatbot = SchoolChatbot()

    """
    Test out generating some responses from the chatbot.
    Inference time
    """
    test_question = f"I'm looking for schools my child is eligible to attend. Our address is {address}. What schools can he attend?"

    print(f"\nQuestion: {test_question}")
    response = chatbot.get_response(test_question)
    print(f"Response: {response}")
    data["Address"].append(address)
    data["Result"].append(response)

results = pd.DataFrame(data)
results.to_csv("location_results.csv", index=False)
```
